```python
!pip install networkx pandas requests
import networkx as nx
import pandas as pd
import requests

API_KEY = "1c38237e2b76ae05ea9a6236a22f92cc"
BASE_URL = "http://ws.audioscrobbler.com/2.0/"

seed_artist = "Tarkan"


def get_similar_artists(artist_name, limit=20):
  params = {
      "method": "artist.getsimilar",
      "artist": artist_name,
      "api_key": API_KEY,
      "format": "json",
      "limit": limit,
  }
  try:
    response = requests.get(BASE_URL, params=params)
    data = response.json()
    return data.get("similarartists", {}).get("artist", [])
  except Exception as e:
    return []


G = nx.Graph()
visited = set()
queue = [seed_artist]

print("Türk müziği benzer sanatçı ağı verisi toplanıyor...")
while queue and len(G.nodes) < 300:
  current_artist = queue.pop(0)
  if current_artist in visited:
    continue
  visited.add(current_artist)

  similar_list = get_similar_artists(current_artist, limit=10)
  for item in similar_list:
    neighbor = item["name"]
    score = float(item.get("match", 0))
    G.add_edge(current_artist, neighbor, weight=score)

    if neighbor not in visited and neighbor not in queue:
      queue.append(neighbor)

print("\n--- ARA KONTROL TEMEL İSTATİSTİKLERİ ---")
print(f"Düğüm (Sanatçı) Sayısı: {G.number_of_nodes()}")
print(f"Kenar (Bağlantı) Sayısı: {G.number_of_edges()}")
print(f"Ağ Yoğunluğu (Density): {nx.density(G):.4f}")
print(
    f"Bağlantılı Bileşen Sayısı (Connected Components):"
    f" {nx.number_connected_components(G)}"
)

# CSV Kaydetme
edgelist_df = nx.to_pandas_edgelist(G)
edgelist_df.to_csv("turk_muzigi_kenar_listesi.csv", index=False)
print(
    "\nKenar listesi 'turk_muzigi_kenar_listesi.csv' olarak başarıyla"
    " oluşturuldu ve kaydedildi!"
)
```
